# How to Build a Successful NBA Team, Based on the Data

**Project type:** Data Science project
**Author:** Karim Elawadi

This document answers one specific question: if I were starting or running an NBA franchise, what should I do based strictly on what the data in this project shows — no general basketball opinions, no assumptions outside the analysis. Every point below is tied directly to a result produced in the notebook's SQL queries, correlations, and charts.

## 1. Do Not Assume Spending More Money Wins More Games

The correlation between total team salary and win rate across all 30 teams was only **0.31** — a weak positive relationship.

- The highest-spending team (Milwaukee Bucks, $122.1M) had a middling win rate of 52.1%.
- The team with the best win rate in the league (San Antonio Spurs, 61.5%) spent a modest $48.1M — less than half the top spender's payroll.
- The lowest-spending team (Detroit Pistons, $7.7M) still competed in the same league as the highest spenders.

**Data-based implication**: A payroll strategy built purely around outspending competitors is not supported by this data. Budget should be allocated toward specific value, not toward total spend as a strategy in itself.

## 2. Prioritize Roster Experience Over Roster Size of Spend

The data showed win rate and average roster age tend to move together: teams with older, more experienced rosters generally showed higher win rates. At the same time, salary rank showed no strong or consistent relationship with either win rate or roster age.

**Data-based implication**: Building a roster around experienced players is more strongly associated with winning than simply having a high payroll. Age/experience appears to be a better data-supported lever than salary size.

## 3. Sign or Draft for Value, Not Just for Raw Stats or Contract Size

The "value-for-money" analysis (points, assists, and rebounds relative to salary) showed that several of the best value contracts in the dataset belonged to role players on minimum or near-minimum salaries (for example, LaMarcus Aldridge and JR Smith as established veterans producing well above their pay level).

**Data-based implication**: A data-driven front office should systematically screen for players whose statistical output exceeds their salary, rather than assuming a bigger contract means a better player. This requires careful data cleaning first — the project found placeholder salary values (identical figures shared across many unrelated players, particularly Two-Way contracts) that would distort this analysis if not filtered out.

## 4. Use Draft Position as a Performance Signal, But Not an All-Star Signal

- Draft pick position showed a moderate negative correlation with overall performance (**-0.41**): top-5 picks averaged a performance score of 25.6, declining steadily to 12.6 for picks 31-60.
- However, draft pick position was **not** meaningfully related to All-Star selection specifically — the correlation between pick number and All-Star appearances was only **0.022**, essentially negligible.

**Data-based implication**: Early draft picks are, on average, a reasonable bet for solid contributors, but draft position alone should not be used to predict star-level talent. Star potential in this dataset was effectively independent of where a player was picked.

## 5. Weigh Physical Attributes Only for Rebounding, Not for Scoring or Playmaking

Height, weight, and wingspan each showed a moderate positive correlation with rebounding (around 0.42-0.43), but a weak, slightly negative correlation with points and assists.

**Data-based implication**: Physical measurements are a legitimate, data-supported factor when evaluating rebounding ability specifically. They are not a data-supported basis for judging scoring or playmaking potential — using size as a general talent filter is not backed by this dataset.

## 6. Exploit Home-Court Advantage Deliberately — It Is Real and Consistent

Across 101 referees with at least 100 games officiated, home teams won **58.8%** of games on average, with every referee's rate sitting well above the 50% break-even point except a small handful near the low end. This confirms home-court advantage is a real, stable effect and not driven by any particular referee or team-referee pairing.

**Data-based implication**: Scheduling, arena environment, and travel planning for opponents are levers with a proven data-backed payoff. There is no evidence in this data of referee bias to exploit or worry about — the advantage comes from being at home itself, not from officiating.

## 7. Track Team Form Separately for Home Games and Away Games

When a team's recent performance ("form" over its last 10 games) was calculated using all games mixed together, its correlation with winning the next game was 0.28-0.31. When form was recalculated separately — a team's form in its own past home games used only when it is playing at home, and its form in its own past away games used only when it is playing away — the correlation rose to 0.33-0.34.

**Data-based implication**: A team's readiness should be assessed in the context of the specific game it is about to play (home vs. away), not from a blended average of all recent games. This is a small but real and repeatable improvement in signal.

## 8. Rest Days Alone Did Not Show a Useful Signal in This Data

The difference in rest days between the two teams before a game showed a correlation of only **0.021** with the outcome — effectively no linear relationship.

**Data-based implication**: A simple day-count of rest did not predict outcomes in this dataset. This does not rule out fatigue effects in general — it means this particular way of measuring rest was not useful here, and a different formulation (for example, flagging true back-to-back games specifically) would need to be tested before relying on it.

## Summary: A Data-Driven Priority Order

Based strictly on the correlations and comparisons produced in this project, if building a competitive franchise from the data alone:

1. Build for home-court advantage — it is the single most consistent and reliable effect found (58.8% win rate).
2. Prioritize roster experience over payroll size.
3. Screen for value contracts (output relative to salary), after cleaning out placeholder/invalid salary data.
4. Use draft position as a floor-performance signal, not a star-potential signal.
5. Use physical attributes only to evaluate rebounding, not scoring or playmaking.
6. Track team form in a context-specific way (home form vs. away form), not as one blended number.
7. Do not over-invest in payroll size or a simple rest-days count as strategic levers — neither showed a strong, reliable relationship with winning in this data.

## Limitations of This Conclusion

This is a correlational analysis of a single dataset, not a causal study and not a complete model of what makes a franchise successful. It does not account for coaching quality, team chemistry, injuries beyond what was already in the tables, in-season trades, scouting depth, or any factor not present in the underlying database. The relationships reported here (0.31, 0.34, 0.41, and so on) are moderate at best — useful as directional evidence, not as guarantees.

# Referee Analysis: Is Home-Court Advantage Driven by Officiating Bias?

**Project type:** Data Science project
**Author:** Karim Elawadi

This document answers one specific question: does referee assignment influence home-court advantage in the NBA, or is the advantage independent of who officiates the game? The analysis is based strictly on the `Game_Officials` and `Game` tables in this project's dataset.

## Method

For each referee with at least 100 officiated games, the home team's win rate was calculated across all games that referee worked. This threshold filters out referees with too few games to draw a reliable conclusion, and avoids small-sample noise skewing the result.

## Key Finding: Home-Court Advantage Is Consistent Across Referees

Across **101 referees** who met the 100-game threshold, the home team won **58.8%** of games on average. Almost every referee's individual home-win rate sat well above the 50% break-even line, with only a small handful of referees near the low end of the range.

**Data-based implication**: If specific referees were biased toward home teams, we would expect to see large variation between referees — some far above 58.8%, some far below, or close to 50%. Instead, the rate is remarkably stable across nearly all 101 officials. This means the home-court advantage is a structural effect (crowd, travel, familiarity with the arena, etc.), **not** an artifact of any particular referee or referee-team pairing.

## What This Rules Out

- There is **no evidence in this data** of individual referee bias favoring home teams.
- There is **no evidence** of specific referees being unusually "tough" or "lenient" on visiting teams in a way that shows up in win rates.
- The home advantage cannot be explained away by "it's just because of who officiated" — it holds regardless of the referee.

## Practical Implication

Because the effect is consistent and not referee-dependent, home-court advantage is a **reliable lever** teams can plan around (scheduling, travel management for opponents, arena environment) — but referee assignment is not something to scrutinize or exploit as part of that strategy.

## Limitation

This is a correlational analysis based on aggregate win rates per referee, not a play-by-play or foul-differential audit. It only speaks to whether *who officiates* is associated with *who wins* — it does not rule out subtler forms of officiating bias (e.g., foul-call differentials) that wouldn't necessarily move the overall win rate.

````python
team_performance = pd.read_sql("""
    SELECT
        TEAM_ID,
        COUNT(*) AS total_games,
        SUM(WIN) AS wins,
        AVG(PTS) AS avg_points
    FROM (
        -- Home games: each team when playing at its own arena
        SELECT
            TEAM_ID_HOME AS TEAM_ID,
            PTS_HOME AS PTS,
            CASE WHEN PTS_HOME > PTS_AWAY THEN 1 ELSE 0 END AS WIN
        FROM Game

        UNION ALL

        -- Away games: the same team when playing at the opponent's arena
        -- (this was previously missing - only home games were counted, away games were ignored)
        SELECT
            VISITOR_TEAM_ID AS TEAM_ID,
            PTS_AWAY AS PTS,
            CASE WHEN PTS_AWAY > PTS_HOME THEN 1 ELSE 0 END AS WIN
        FROM Game
    ) all_games
    GROUP BY TEAM_ID;
""", conn)
team_performance
````

````python
arenas = pd.read_sql("""
    SELECT ta.ID as TEAM_ID, t.full_name, ta.NICKNAME, ta.ARENA, ta.ARENACAPACITY
    FROM Team_Attributes ta
    JOIN Team t ON ta.ID = t.id;
""", conn)
final = pd.merge(team_performance, arenas, on="TEAM_ID", how="inner")
final["win_rate"] = final["wins"] / final["total_games"]
final_sorted = final[["full_name", "ARENACAPACITY", "total_games", "wins", "win_rate", "avg_points"]].sort_values("win_rate", ascending=False)

print(len(final_sorted))
final_sorted
````

````python
correlation = final_sorted["ARENACAPACITY"].corr(final_sorted["win_rate"])
print(f"Correlation between arena capacity and win rate: {correlation}")
````

````python
player_salary_perf = pd.read_sql("""
    SELECT ps.namePlayer, ps.value as salary, pa.PTS, pa.AST, pa.REB, pa.ALL_STAR_APPEARANCES
    FROM Player_Salary ps
    JOIN Player_Attributes pa ON ps.namePlayer = pa.DISPLAY_FIRST_LAST
    WHERE ps.slugSeason = '2023-24' AND pa.PTS IS NOT NULL;
""", conn)

player_salary_perf
````

````python
# ---------- 0) Build a "raw" salary source from Player_Salary, aggregated by team ----------
# This gives us salary_y (the corrected value) to compare against salary_x (the old value from Team_Salary)
team_salary_raw = pd.read_sql("""
    SELECT pa.TEAM_ID, SUM(ps.value) AS salary_raw
    FROM Player_Salary ps
    JOIN Player_Attributes pa ON ps.namePlayer = pa.DISPLAY_FIRST_LAST
    WHERE ps.slugSeason = '2023-24' AND pa.TEAM_ID IS NOT NULL
    GROUP BY pa.TEAM_ID;
""", conn)

team_salary_raw["TEAM_ID"] = team_salary_raw["TEAM_ID"].astype(int)
final2["TEAM_ID"] = final2["TEAM_ID"].astype(int)

# Merge the old (Team_Salary) and new (Player_Salary) sources into one table
final2_fixed = pd.merge(final2, team_salary_raw, on="TEAM_ID", how="left")
final2_fixed = final2_fixed.rename(columns={"salary": "salary_x", "salary_raw": "salary_y"})

print(f"final2_fixed ready: {len(final2_fixed)} rows")
````

````python
# ---------- 1) Compare the two salary sources for the three suspect teams ----------
suspects = ["Chicago Bulls", "New York Knicks", "Orlando Magic"]
comparison = final2_fixed[final2_fixed["full_name"].isin(suspects)][
    ["full_name", "salary_x", "salary_y"]
]
print("Comparison: salary_x (old) vs salary_y (corrected, from source data):")
print(comparison)

# Expected:
# Chicago Bulls   -> salary_x = 0.0   | salary_y = 0.0   (a genuine gap in the source data, stays zero)
# New York Knicks -> salary_x = 0.0   | salary_y = 0.0   (same situation)
# Orlando Magic   -> salary_x = 0.0   | salary_y = 34400000.0  (this is the real discrepancy - previously wrong)

# ---------- 2) Show all teams where salary_x and salary_y differ ----------
diffs = final2_fixed[final2_fixed["salary_x"] != final2_fixed["salary_y"]][
    ["full_name", "salary_x", "salary_y"]
]
print("\nAll teams with a difference between the two versions:")
print(diffs)

# ---------- 3) Adopt the corrected version as final, and drop the old one ----------
final2_fixed["salary"] = final2_fixed["salary_y"]
final2_fixed = final2_fixed.drop(columns=["salary_x", "salary_y"])

# ---------- 4) Replace genuine missing-data zeros (Bulls, Knicks) with NaN instead of leaving them at zero ----------
# Because zero here means "missing data", not "an actual salary of zero"
final2_fixed.loc[final2_fixed["salary"] == 0, "salary"] = pd.NA

# ---------- 5) Now compute the correlation heatmap on the actually-corrected data ----------
cols = ["ARENACAPACITY", "salary", "avg_age", "total_games", "wins", "win_rate", "avg_points"]
correlation_table_final = final2_fixed[cols].corr()
print("\nCorrelation matrix after all corrections:")
print(correlation_table_final.round(2))


# ---------- 6) Propagate the corrected salaries back into final2 so all charts and the map use the correct numbers ----------
final2 = final2.merge(
    final2_fixed[["TEAM_ID", "salary"]].rename(columns={"salary": "salary_corrected"}),
    on="TEAM_ID", how="left"
)
final2["salary"] = final2["salary_corrected"]
final2 = final2.drop(columns=["salary_corrected"])

print("\nfinal2 updated with corrected salaries - all charts and the map will now use the correct numbers")
````

## Chart 1: Points Distribution by Salary Tier

This chart visualizes the relationship between NBA players' salaries and their
scoring performance for the 2023-24 season.

Player salary and points-per-game (PTS) data were pulled from the database and
split into three equal-sized groups using quantile-based binning (`qcut`):
**Low**, **Medium**, and **High** salary tiers. A stacked histogram then shows

## 1. Top Referees by Wins — Bar Chart Analysis

**How to read it:** compare referees by win count and look for any official

- **Color:** Higher values indicate more wins for that team-referee pairing

**How to read it:** look for any team-referee combination that is unusually
high or low relative to the rest of the grid, which could indicate a pairing
worth investigating further.

---

## 3. Top Referee-Team Winning Combinations — Grouped Bar Analysis

This chart plots the same top-5-teams-by-top-10-referees data used in the
heatmap above (section 2), as grouped bars instead of a color grid. Each
referee's wins are grouped along the x-axis, with one colored bar per team
that referee actually officiated a win for - a referee who only appears with
a single team from the top 5 shows one bar, while a referee who appears with
several teams shows all of those bars clustered together under their name.

**How to read it:** grouped bars (instead of stacked) allow a fair visual
comparison between referees and teams, and make it easy to spot any single
referee-team pairing that dominates the others. Since this chart is built
from the same underlying numbers as the heatmap in section 2, the two should
always agree - any bar here matches the corresponding cell value there.

---

## 4. Home Team Win Percentage by Referee — Bar Chart Analysis

This chart ranks referees by the percentage of games won by the home team.

**How to read it:** a consistent home win percentage across referees

referee-driven bias in home-court outcomes.

## 5. Referee Impact on Home Team Wins — Scatter Plot Analysis

This chart examines the relationship between the number of games officiated and the home team win percentage.

**Chart Encoding:**
- **X-axis:** Number of games officiated

- **Bubble size:** Total games officiated
- **Color:** Home team win percentage

**How to read it:** referees with fewer games officiated tend to show more

home-court advantage independent of who is officiating.

````python
query = """
WITH winners AS (

    SELECT
        GAME_ID,
        HOME_TEAM_ID AS TEAM_ID
    FROM Game
    WHERE PTS_HOME > PTS_AWAY

    UNION ALL

    SELECT
        GAME_ID,
        VISITOR_TEAM_ID AS TEAM_ID
    FROM Game
    WHERE PTS_AWAY > PTS_HOME
)

SELECT
    t.full_name AS TEAM,
    go.FIRST_NAME || ' ' || go.LAST_NAME AS REFEREE
FROM winners w
JOIN Team t
    ON w.TEAM_ID = t.id
JOIN Game_Officials go
    ON w.GAME_ID = go.GAME_ID
"""

df = pd.read_sql(query, conn)

top5 = df['TEAM'].value_counts().head(5).index

df_top = df[df['TEAM'].isin(top5)]

relation = (
    df_top.groupby(['TEAM', 'REFEREE'])
          .size()
          .reset_index(name='WINS')
)

relation.sort_values('WINS', ascending=False).head(20)
````

````python
# `pivot`: top 5 teams x top 10 referees, used by the heatmap below.
top10_refs = (
    relation.groupby('REFEREE')['WINS'].sum()
    .sort_values(ascending=False).head(10).index
)

pivot = (
    relation[relation['REFEREE'].isin(top10_refs)]
    .pivot_table(index='TEAM', columns='REFEREE', values='WINS', fill_value=0)
)

# `heat`: the real referee-team combinations only (no zero-filled bars) for
# the top 10 referees, grouped and ordered by each referee's total wins so
# the bar chart below groups every referee with all of the top-5 teams they
# actually officiated a win for.
heat = relation[relation['REFEREE'].isin(top10_refs)].copy()

referee_order = (
    heat.groupby('REFEREE')['WINS'].sum().sort_values(ascending=False).index.tolist()
)
heat['REFEREE'] = pd.Categorical(heat['REFEREE'], categories=referee_order, ordered=True)
heat = heat.sort_values(['REFEREE', 'TEAM'])


query_state = """
WITH winners AS (
    SELECT GAME_ID, HOME_TEAM_ID AS TEAM_ID FROM Game WHERE PTS_HOME > PTS_AWAY
    UNION ALL
    SELECT GAME_ID, VISITOR_TEAM_ID AS TEAM_ID FROM Game WHERE PTS_AWAY > PTS_HOME
)
SELECT
    t.full_name AS TEAM,
    t.state AS STATE_NAME,
    go.FIRST_NAME || ' ' || go.LAST_NAME AS REFEREE
FROM winners w
JOIN Team t ON w.TEAM_ID = t.id
JOIN Game_Officials go ON w.GAME_ID = go.GAME_ID
"""
df_all_teams = pd.read_sql(query_state, conn)


state_abbr_map = {name: info[0] for name, info in state_centroids.items()}
df_all_teams["STATE"] = df_all_teams["STATE_NAME"].map(state_abbr_map)
df_all_teams = df_all_teams.dropna(subset=["STATE"])   

heat2 = (
    df_all_teams.groupby(['TEAM', 'STATE', 'REFEREE'])
    .size()
    .reset_index(name='WINS')
)

print(f"heat: {len(heat)} rows | pivot: {pivot.shape} | heat2: {len(heat2)} rows")
````

- contract value, to identify players whose output exceeds their salary
  (while checking for unusually low salary values that can distort this metric), and
- physical attributes, considered only in the appropriate context (e.g., rebounding),

rather than relying on draft pedigree or physical size alone.

````python
from IPython.display import display

# Correction notes (found after running against the real database):
# Filtering player_salary to season 2023-24 only shrank the sample too much
# (only currently active players), which caused ALL_STAR_APPEARANCES to come
# back as NaN across the board in correlation_pick (not enough variance in
# such a small sample).
# Fix: use each player's most recently recorded salary (not necessarily
# 2023-24 specifically) - this keeps a wider sample (including older/retired
# draftees) while still using a single value per player (rather than
# averaging across a player's whole career, which was the original bug).
# Also, ALL_STAR_APPEARANCES being NaN more accurately means "zero
# appearances" rather than "missing data".
query = """
WITH latest_salary AS (
    SELECT
        namePlayer,
        value,
        slugSeason,
        ROW_NUMBER() OVER (PARTITION BY namePlayer ORDER BY slugSeason DESC) AS rn
    FROM Player_Salary
)
SELECT D.idPlayer, D.namePlayer, D.yearDraft, D.numberPickOverall, D.teamName,
       S.value AS avg_salary, S.slugSeason AS salary_season,
       A.PTS, A.AST, A.REB,
       COALESCE(A.ALL_STAR_APPEARANCES, 0) AS ALL_STAR_APPEARANCES,
       DC.heightWOShoesInches, DC.weightLBS, DC.wingspanInches
FROM Draft AS D
INNER JOIN latest_salary AS S
    ON D.namePlayer = S.namePlayer AND S.rn = 1
INNER JOIN Player_Attributes AS A
    ON A.ID = D.idPlayer
INNER JOIN Draft_Combine AS DC
    ON DC.idPlayer = D.idPlayer
"""
df = pd.read_sql(query, conn)

# A salary of zero or missing means missing data, not an actual salary of zero -
# leaving it as-is would make value_score come back as infinite/inflated and misleading.
df.loc[df['avg_salary'] <= 0, 'avg_salary'] = pd.NA

# Some salary values are shared identically across many different players
# (e.g. 110998.0 appears for 14 different players, 50000.0 for 2 different
# players, all in season 2020-21) - these are placeholder/estimated values
# in the source data (likely for Two-Way contracts), not real individual
# salaries. Treat any salary shared by more than one player as missing
# rather than using it in value_score, since dividing by a fake salary
# produces a fake (usually inflated) value_score.
salary_counts = df['avg_salary'].value_counts()
shared_salaries = salary_counts[salary_counts > 1].index
df.loc[df['avg_salary'].isin(shared_salaries), 'avg_salary'] = pd.NA

# Value-for-money score: performance divided by salary (in millions)
df['value_score'] = (df['PTS'] + df['AST'] + df['REB']) / (df['avg_salary'] / 1_000_000)

print("Top 10 players by value-for-money:")
display(df.sort_values('value_score', ascending=False).head(10)[
    ['namePlayer', 'teamName', 'salary_season', 'avg_salary', 'PTS', 'AST', 'REB', 'value_score']
])

# Correlation between physical attributes and performance
correlation = df[['heightWOShoesInches', 'weightLBS', 'wingspanInches',
                   'PTS', 'AST', 'REB']].corr()
print("\nCorrelation between physical attributes and performance:")
display(correlation[['PTS', 'AST', 'REB']])

# Correlation between draft pick number and performance
df['performance_total'] = df['PTS'] + df['AST'] + df['REB']

correlation_pick = df[['numberPickOverall', 'performance_total',
                        'ALL_STAR_APPEARANCES']].corr()
print("\nCorrelation between draft position and performance:")
display(correlation_pick)
````

##  Data Quality Notes — Columns to Watch

Before reusing this dataset elsewhere in the project, keep these points in mind:

### `Player_Salary`
- **`value`** — some salary values are shared identically across many different players (placeholder values, not real salaries). Check with `value_counts()` and treat repeated values as missing.
- **`namePlayer`** — no unique ID. Don't join on this column; use `idPlayer` where available.
- **`slugSeason`** — not every player has a record in every season. Filtering to one season can silently drop active players.

### `Player_Attributes`
- **`TEAM_ID`** — reflects the player's most recent/any team across their whole career, not necessarily their current-season team.
- **`ALL_STAR_APPEARANCES`** — returns `NaN` for most players instead of `0`. Use `COALESCE(..., 0)` before any calculation.
- **`DISPLAY_FIRST_LAST`** — same name-only risk as `namePlayer`. Use `ID` for joins instead.

### `Team_Salary`
- **Season salary column (e.g. `X2023-24`)** — a value of `0` usually means missing data, not an actual zero salary. Convert to `NaN` first.
- **`nameTeam`** — must match `Team.full_name` exactly for joins to work correctly.

### `Game`
- **`TEAM_ID_HOME`** / **`VISITOR_TEAM_ID`** — using only one gives half the data (only home or only away games). Combine both with `UNION ALL` for accurate win rates.

### `Draft`
- **`namePlayer`** — same name-collision risk. Use `idPlayer` for joins.

---

**General rule of thumb:**
1. Never join on a name — use an ID column.
2. `0` or `NaN` doesn't always mean "no data" — it can be a placeholder or missing-data marker.
3. Filtering to a single season can drop real, active players.
4. Using only one side of a home/away pair gives incomplete results.

## 1. Win Rate Calculation Used Home Games Only

**Original issue**

Win rate for each team was computed using only games where the team appeared as the home team (`TEAM_ID_HOME`), ignoring games played as the visiting team entirely.

**Why it was wrong**

Every team plays roughly half its games at home and half away. Computing win rate from only half the games produces a figure that can diverge significantly from the team's actual season performance.

**Fix**

A `UNION ALL` was used to combine each team's home-game and away-game perspective into a single table before computing win rate:

```sql
SELECT TEAM_ID_HOME AS TEAM_ID, PTS_HOME AS PTS,
       CASE WHEN PTS_HOME > PTS_AWAY THEN 1 ELSE 0 END AS WIN
FROM Game
UNION ALL
SELECT VISITOR_TEAM_ID AS TEAM_ID, PTS_AWAY AS PTS,
       CASE WHEN PTS_AWAY > PTS_HOME THEN 1 ELSE 0 END AS WIN
FROM Game
```

**Verification**

The first five home games for the San Antonio Spurs were checked manually against `PTS_HOME` and `PTS_AWAY`, and the computed `WIN` value matched in every case.

# Feature Engineering Explanation: Team Form Features

This section builds "recent form" features for each team (based on their last 10 games) and then merges them into a single game-level dataset used for prediction.

## Step 1: Building the `team_games` Table

```python
team_games = pd.read_sql("""
    SELECT GAME_ID, GAME_DATE, TEAM_ID_HOME AS TEAM_ID, PTS_HOME AS PTS, PTS_AWAY AS OPP_PTS,
           CASE WHEN PTS_HOME > PTS_AWAY THEN 1 ELSE 0 END AS WIN
    FROM Game
    UNION ALL
    SELECT GAME_ID, GAME_DATE, VISITOR_TEAM_ID AS TEAM_ID, PTS_AWAY AS PTS, PTS_HOME AS OPP_PTS,
           CASE WHEN PTS_AWAY > PTS_HOME THEN 1 ELSE 0 END AS WIN
    FROM Game
""", conn)
```

- The original `Game` table stores one row per game, with separate home and away columns.
- This query uses `UNION ALL` to turn that into **one row per team per game** instead of one row per game. Each team (whether it was home or away) gets its own row with its own points scored (`PTS`), points allowed (`OPP_PTS`), and whether it won (`WIN`).
- This "long format" makes it much easier to calculate a team's own rolling statistics, since all of a team's games (home and away combined) are now in a single column grouped by `TEAM_ID`.

Additional columns are then computed:
- `GAME_DATE` is converted to an actual datetime type.
- `MARGIN` = `PTS - OPP_PTS` (point differential for that team in that game).
- The table is sorted by `TEAM_ID` and `GAME_DATE` so that rolling calculations happen in chronological order per team.

## Step 2: Rolling "Last 10 Games" Features

```python
team_games["L10_WIN_RATE"] = (
    team_games.groupby("TEAM_ID")["WIN"]
    .transform(lambda s: s.shift(1).rolling(10, min_periods=5).mean())
)

team_games["L10_AVG_MARGIN"] = (
    team_games.groupby("TEAM_ID")["MARGIN"]
    .transform(lambda s: s.shift(1).rolling(10, min_periods=5).mean())
)
```

- For each team, this computes:
  - `L10_WIN_RATE`: the team's win rate over its last 10 games.
  - `L10_AVG_MARGIN`: the team's average point margin over its last 10 games.
- The key detail is `.shift(1)` before `.rolling(10)`. This shifts the data forward by one game so that **the current game's result is excluded** from its own rolling average. Without this shift, the model would be "leaking" the outcome of the game it's trying to predict into its own features.
- `min_periods=5` means a team needs at least 5 prior games before a rolling average is calculated; otherwise the value is `NaN` (visible in the early rows of the output, e.g. a team's first few games of the 1949 season).

At this point, `team_games` has one row per team per game, with that team's rolling win rate and average margin heading into the game.

## Step 3: Rebuilding the Game-Level Dataset

```python
games = pd.read_sql("""
    SELECT GAME_ID, GAME_DATE, TEAM_ID_HOME, VISITOR_TEAM_ID,
           CASE WHEN PTS_HOME > PTS_AWAY THEN 1 ELSE 0 END AS HOME_WIN
    FROM Game
""", conn)
```

- This re-reads the original one-row-per-game table, but now just needs the IDs and a `HOME_WIN` label (1 if the home team won, 0 otherwise). This is the target variable.

```python
home_form = team_games[["GAME_ID", "TEAM_ID", "L10_WIN_RATE", "L10_AVG_MARGIN"]].rename(
    columns={"TEAM_ID": "TEAM_ID_HOME", "L10_WIN_RATE": "HOME_L10_WIN_RATE", "L10_AVG_MARGIN": "HOME_L10_MARGIN"}
)

away_form = team_games[["GAME_ID", "TEAM_ID", "L10_WIN_RATE", "L10_AVG_MARGIN"]].rename(
    columns={"TEAM_ID": "VISITOR_TEAM_ID", "L10_WIN_RATE": "AWAY_L10_WIN_RATE", "L10_AVG_MARGIN": "AWAY_L10_MARGIN"}
)
```

- The long-format `team_games` table (one row per team per game) is split back into two views:
  - `home_form`: the rolling stats renamed as if that team were the home team.
  - `away_form`: the rolling stats renamed as if that team were the away team.
- This renaming is what allows each row of the final dataset to carry both teams' form stats side by side.

```python
dataset = games.merge(home_form, on=["GAME_ID", "TEAM_ID_HOME"]).merge(away_form, on=["GAME_ID", "VISITOR_TEAM_ID"])

dataset["FORM_DIFF_WIN_RATE"] = dataset["HOME_L10_WIN_RATE"] - dataset["AWAY_L10_WIN_RATE"]
dataset["FORM_DIFF_MARGIN"] = dataset["HOME_L10_MARGIN"] - dataset["AWAY_L10_MARGIN"]

dataset = dataset.dropna()
```

- The two form tables are merged back onto the game-level `games` table, matching on `GAME_ID` and the respective team ID.
- Two new "differential" features are created:
  - `FORM_DIFF_WIN_RATE`: how much better (or worse) the home team's recent win rate is compared to the away team's.
  - `FORM_DIFF_MARGIN`: same idea, but for average point margin.
- `dropna()` removes early-season games where a team didn't yet have 5 prior games to compute a rolling average (the `NaN` rows seen earlier).

The final `dataset` has 62,691 rows and 11 columns: game identifiers, each team's rolling form stats, the two differential features, and the `HOME_WIN` label.

## Step 4: Checking Correlation with the Target

```python
print(dataset[["HOME_WIN", "FORM_DIFF_WIN_RATE", "FORM_DIFF_MARGIN"]].corr())
```

This checks how well the new features relate to the outcome being predicted (`HOME_WIN`):

| | HOME_WIN | FORM_DIFF_WIN_RATE | FORM_DIFF_MARGIN |
|---|---|---|---|
| HOME_WIN | 1.000 | 0.285 | 0.309 |
| FORM_DIFF_WIN_RATE | 0.285 | 1.000 | 0.884 |
| FORM_DIFF_MARGIN | 0.309 | 1.000 | 0.884 |

- Both `FORM_DIFF_WIN_RATE` (0.285) and `FORM_DIFF_MARGIN` (0.309) have a positive but moderate correlation with `HOME_WIN`. This means teams that have been performing better recently do tend to win more, but recent form alone doesn't fully determine the outcome (home-court advantage, matchups, injuries, etc. also matter).
- `FORM_DIFF_WIN_RATE` and `FORM_DIFF_MARGIN` are highly correlated with each other (0.884), which makes sense since a team that wins more games also tends to win by larger margins. This is worth keeping in mind if both are used together in a model, since they carry overlapping information (multicollinearity).

## Summary

This code takes a raw game-by-game database table and engineers "recent team form" features (rolling win rate and point margin over each team's last 10 games, properly shifted to avoid data leakage), then combines the home and away teams' form into a single differential feature set that can be used to predict which team wins a given game.

````python
# Combine home and away rows into one team-game table, ordered by date
team_games = pd.read_sql("""
    SELECT GAME_ID, GAME_DATE, TEAM_ID_HOME AS TEAM_ID, PTS_HOME AS PTS, PTS_AWAY AS OPP_PTS,
           CASE WHEN PTS_HOME > PTS_AWAY THEN 1 ELSE 0 END AS WIN
    FROM Game
    UNION ALL
    SELECT GAME_ID, GAME_DATE, VISITOR_TEAM_ID AS TEAM_ID, PTS_AWAY AS PTS, PTS_HOME AS OPP_PTS,
           CASE WHEN PTS_AWAY > PTS_HOME THEN 1 ELSE 0 END AS WIN
    FROM Game
""", conn)

team_games["GAME_DATE"] = pd.to_datetime(team_games["GAME_DATE"])
team_games["MARGIN"] = team_games["PTS"] - team_games["OPP_PTS"]
team_games = team_games.sort_values(["TEAM_ID", "GAME_DATE"])

# Rolling window over the PRIOR 10 games only (shift(1) excludes the current game)
team_games["L10_WIN_RATE"] = (
    team_games.groupby("TEAM_ID")["WIN"]
    .transform(lambda s: s.shift(1).rolling(10, min_periods=5).mean())
)
team_games["L10_AVG_MARGIN"] = (
    team_games.groupby("TEAM_ID")["MARGIN"]
    .transform(lambda s: s.shift(1).rolling(10, min_periods=5).mean())
)
team_games
````

## Extending the Model — Rest Days & Home/Away-Specific Form

The previous model only used **overall** L10 win rate and margin (mixing home and away games together). Two improvements:

1. **Rest days**: number of days since each team's previous game (captures back-to-back fatigue).
2. **Home/away-specific form**: instead of a team's *overall* L10 form, use its form **specifically in home games** (for the home team) and **specifically in away games** (for the away team) — this better reflects how a team performs in the exact context of the upcoming game.


````python
# --- Rest days: days since each team's previous game ---
team_games["REST_DAYS"] = (
    team_games.groupby("TEAM_ID")["GAME_DATE"]
    .diff()
    .dt.days
)
# First game of the season for a team has no prior game -> fill with a neutral value (e.g. 3 days)
team_games["REST_DAYS"] = team_games["REST_DAYS"].fillna(3)
# Cap extreme values (long All-Star break, offseason boundary rows, etc.)
team_games["REST_DAYS"] = team_games["REST_DAYS"].clip(upper=7)

# --- Home/away-specific rolling form (context-matched instead of overall L10) ---
team_games = team_games.sort_values(["TEAM_ID", "GAME_DATE"])

# Home-context rolling form: only using the team's own past HOME games
team_games["IS_HOME_ROW"] = None  # placeholder, set from the two source queries below
````

````python
# Rebuild team_games with a HOME/AWAY flag per row so we can compute context-specific rolling form
team_games_ctx = pd.read_sql("""
    SELECT GAME_ID, GAME_DATE, TEAM_ID_HOME AS TEAM_ID, PTS_HOME AS PTS, PTS_AWAY AS OPP_PTS,
           CASE WHEN PTS_HOME > PTS_AWAY THEN 1 ELSE 0 END AS WIN,
           1 AS IS_HOME
    FROM Game
    UNION ALL
    SELECT GAME_ID, GAME_DATE, VISITOR_TEAM_ID AS TEAM_ID, PTS_AWAY AS PTS, PTS_HOME AS OPP_PTS,
           CASE WHEN PTS_AWAY > PTS_HOME THEN 1 ELSE 0 END AS WIN,
           0 AS IS_HOME
    FROM Game
""", conn)

team_games_ctx["GAME_DATE"] = pd.to_datetime(team_games_ctx["GAME_DATE"])
team_games_ctx["MARGIN"] = team_games_ctx["PTS"] - team_games_ctx["OPP_PTS"]
team_games_ctx = team_games_ctx.sort_values(["TEAM_ID", "GAME_DATE"])

# Rest days (same logic as above, computed on the full per-team timeline)
team_games_ctx["REST_DAYS"] = (
    team_games_ctx.groupby("TEAM_ID")["GAME_DATE"].diff().dt.days
)
team_games_ctx["REST_DAYS"] = team_games_ctx["REST_DAYS"].fillna(3).clip(upper=7)

# Context-specific rolling form: for each team, rolling over ONLY its home rows (or ONLY its away rows)
def add_context_rolling(df, is_home_flag, win_col, margin_col):
    mask = df["IS_HOME"] == is_home_flag
    sub = df.loc[mask].sort_values(["TEAM_ID", "GAME_DATE"])
    sub[win_col] = sub.groupby("TEAM_ID")["WIN"].transform(
        lambda s: s.shift(1).rolling(10, min_periods=3).mean()
    )
    sub[margin_col] = sub.groupby("TEAM_ID")["MARGIN"].transform(
        lambda s: s.shift(1).rolling(10, min_periods=3).mean()
    )
    return sub[["GAME_ID", "TEAM_ID", win_col, margin_col]]

home_ctx_form = add_context_rolling(team_games_ctx, 1, "HOME_CTX_WIN_RATE", "HOME_CTX_MARGIN")
away_ctx_form = add_context_rolling(team_games_ctx, 0, "AWAY_CTX_WIN_RATE", "AWAY_CTX_MARGIN")

rest_days = team_games_ctx[["GAME_ID", "TEAM_ID", "REST_DAYS"]]
````

````python
# --- Rebuild the modeling dataset with the new context-specific + rest-day features ---
games = pd.read_sql("""
    SELECT GAME_ID, GAME_DATE, TEAM_ID_HOME, VISITOR_TEAM_ID,
           CASE WHEN PTS_HOME > PTS_AWAY THEN 1 ELSE 0 END AS HOME_WIN
    FROM Game
""", conn)

# Overall L10 form (kept from before, for comparison)
home_form = team_games[["GAME_ID", "TEAM_ID", "L10_WIN_RATE", "L10_AVG_MARGIN"]].rename(
    columns={"TEAM_ID": "TEAM_ID_HOME", "L10_WIN_RATE": "HOME_L10_WIN_RATE", "L10_AVG_MARGIN": "HOME_L10_MARGIN"}
)
away_form = team_games[["GAME_ID", "TEAM_ID", "L10_WIN_RATE", "L10_AVG_MARGIN"]].rename(
    columns={"TEAM_ID": "VISITOR_TEAM_ID", "L10_WIN_RATE": "AWAY_L10_WIN_RATE", "L10_AVG_MARGIN": "AWAY_L10_MARGIN"}
)

# Context-specific form (new)
home_ctx = home_ctx_form.rename(columns={"TEAM_ID": "TEAM_ID_HOME"})
away_ctx = away_ctx_form.rename(columns={"TEAM_ID": "VISITOR_TEAM_ID"})

# Rest days for each side (new)
home_rest = rest_days.rename(columns={"TEAM_ID": "TEAM_ID_HOME", "REST_DAYS": "HOME_REST_DAYS"})
away_rest = rest_days.rename(columns={"TEAM_ID": "VISITOR_TEAM_ID", "REST_DAYS": "AWAY_REST_DAYS"})

dataset2 = (
    games
    .merge(home_form, on=["GAME_ID", "TEAM_ID_HOME"])
    .merge(away_form, on=["GAME_ID", "VISITOR_TEAM_ID"])
    .merge(home_ctx, on=["GAME_ID", "TEAM_ID_HOME"])
    .merge(away_ctx, on=["GAME_ID", "VISITOR_TEAM_ID"])
    .merge(home_rest, on=["GAME_ID", "TEAM_ID_HOME"])
    .merge(away_rest, on=["GAME_ID", "VISITOR_TEAM_ID"])
)

dataset2["FORM_DIFF_WIN_RATE"] = dataset2["HOME_L10_WIN_RATE"] - dataset2["AWAY_L10_WIN_RATE"]
dataset2["FORM_DIFF_MARGIN"] = dataset2["HOME_L10_MARGIN"] - dataset2["AWAY_L10_MARGIN"]
dataset2["CTX_FORM_DIFF_WIN_RATE"] = dataset2["HOME_CTX_WIN_RATE"] - dataset2["AWAY_CTX_WIN_RATE"]
dataset2["CTX_FORM_DIFF_MARGIN"] = dataset2["HOME_CTX_MARGIN"] - dataset2["AWAY_CTX_MARGIN"]
dataset2["REST_DIFF"] = dataset2["HOME_REST_DAYS"] - dataset2["AWAY_REST_DAYS"]

dataset2 = dataset2.dropna()
print(f"dataset2 rows: {len(dataset2)}")
dataset2[["HOME_WIN", "FORM_DIFF_WIN_RATE", "CTX_FORM_DIFF_WIN_RATE", "CTX_FORM_DIFF_MARGIN", "REST_DIFF"]].corr()
````
